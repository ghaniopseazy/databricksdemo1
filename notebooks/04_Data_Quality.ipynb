# 04 — Simple Data Quality Check

Run this after Silver and before Gold in the Lakeflow Job.


```python
from pyspark.sql import functions as F

orders = spark.table("retail_catalog.silver.orders")

total = orders.count()
bad_ids = orders.filter(F.col("order_id").isNull()).count()
bad_quantity = orders.filter(F.col("quantity") <= 0).count()

print("Total orders:", total)
print("Null order IDs:", bad_ids)
print("Invalid quantities:", bad_quantity)

if total == 0:
    raise Exception("DQ FAILED: Silver orders is empty")

if bad_ids > 0:
    raise Exception("DQ FAILED: Null order IDs")

if bad_quantity > 0:
    raise Exception("DQ FAILED: Invalid quantity")

print("DATA QUALITY PASSED")
```
